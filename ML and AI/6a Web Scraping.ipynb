# Web Scraping

### What is it?

Web scraping is the art and practice of **retrieving information from a website**. Web scraping can be done manually and, in a sense, it's what we do every time we look up some information on the website or when we copy a table from a Wikipedia page and paste it into a spreadsheet.

That said, the most efficient way to scrape a website is through some kind of an automated script and we will explore this practice in Python using a well known scraping library called `BeautifulSoup`.

### How does it work?

1. The first step is to locate a website from which you want to extract some information that is embedded in it. It could be an e-Commerce, a news outlet, a weather site or any other website really.
2. You should then proceed to inspect and analyze the HTML source code of that web page to find the specific ids and tags that identify the information you're after.
3. The central part of a Python scraper is, of course, the script itself. You write an algorithm to instruct your program to use those tags and ids to retrieve the information you're after.
4. Finally, you manipulate your results into a data structure of your choice (csv, json, ...) that will allow you to interact with and use your scraped data.

### Is it legal?

In general terms, web scraping is not illegal. More specifically, according to Wikipedia: *the legality of web scraping varies across the world. In general, web scraping may be against the terms of use of some websites, but the enforceability of these terms is unclear.*

It is generally accepted that, if a website makes its data public, then scraping it should be legal. This is reinforced by legal cases such as "LinkedIn vs hiQ Labs". You can read more on [Wikipedia](https://en.wikipedia.org/wiki/Web_scraping#Legal_issues) and [online](https://www.parsehub.com/blog/web-scraping-legal/).

### A first simple example

```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
```

We will start with a [simple page](https://www.scrapethissite.com/pages/simple/), click on the link and familiarise yourself with its contents; it contains a list of country names from the world and some information for each one of them.

Before moving on, inspect the underlying HTML code: right click on an element of the page (for example the name "Andorra") and select "Inspect". This will open the Inspect Element window in your Chrome browser, which allows you to see the code that generates the page you're looking at.

<img src="images/inspect-element.png" width="600">

In Python, let's start by sending an HTTP GET request to retrieve the HTML content of a desired web page and print to screen the first 1000 characters:

```python
url = "https://www.scrapethissite.com/pages/simple/"
page = requests.get(url)

print(page.text[0:1000])
```

```python
type(page.text)
```

Then we create a Beautiful Soup object named `soup` that takes `page.content` as an input (this is the HTML content we just printed to screen).

*Note: it is better to pass page.content instead of page.text to avoid issues with character encoding.*

```python
soup = BeautifulSoup(page.content, "html.parser")
```

```python
type(soup)
```

If you go back to the **Inspect Element** window, you'll notice that if you hover over an element in the windown on the right, it will highlight on the webpage to the left. We want to retrieve all the country names in the list and the `<div id="page">` element contains all of the results that we're after, so we apply the `.find()` method to the `soup` object in order to find that element via its `id="page"` parameter.

<img src="images/div id=page.png" width="600">

```python
results = soup.find(id="page")
```

```python
results
```

As we were saying, inside the `results` object there are all the country names we're interested in, but we still need to identify them and to do that we go back to the Inspect Element window and notice that each country block has its name and all related information stored inside a `div` element with `class="col-md-4 country"`. Therefore, we apply the `.find_all()` method to the `results` object in order to get the information about each country.

```python
countries = results.find_all("div", class_="col-md-4 country")
```

```python
type(countries)
```

```python
len(countries)
```

```python
countries[0]
```

```python
type(countries[0])
```

Now, if you look closely to the above output, you'll notice that each country name is enclosed in a `h3` tag with `class="country-name"`.

```python
type(countries)
```

```python
len(countries)
```

The output object `countries` is an iterable object of class `bs4.element.ResultSet` containing 250 elements, so we can use a for loop to print each element in the `h3` tag.

```python
countries[0:3]
```

```python
for c in countries[0:3]:
    print(c.find("h3", class_="country-name"))
```

As you can see, this prints everything, including the tag itself; in order **to include just the text**, we can add the `.text` attribute.

```python
for c in countries[0:3]:
    print(c.find("h3", class_="country-name").text)
```

Now, instead of printing the output to screen, it's better to append each new element to a list in order to store those information for a later use.

```python
listy = []
for c in countries:
    listy.append(c.find("h3", class_="country-name").text)
```

```python
listy[0:5]
```

Since there are some "spaces" and "return" characters in the strings, we use the `.strip()` method to remove those from the output.

```python
type(listy[0])
```

```python
names = []
for c in countries:
    names.append(c.find("h3", {"class":"country-name"}).text.strip())
```

```python
names
```

```python
countries[0]
```

```python
int(countries[0].find("span", class_="country-population").text)
```

```python
names[0:5]
```

That's it, we successfully created a list including all the country names from our initial web page.

### A slightly more difficult example

At [this link](https://www.rottentomatoes.com/browse/movies_at_home/affiliates:netflix~sort:critic_highest) you will find a list of the top movies (according to the website's Tomatometer score) currently available on Netflix. We want to retrieve some information (such as the movie title) for all the movies available in the page and save such information in a pandas data frame.

We begin by sending an HTTP GET request to retrieve the HTML page and save its contents to a `BeautifulSoup` object.

*Notice the `headers={'Accept-Language': "lang=en-US"}` additional parameter, which ensures that the contents are always retrieved in english, no matter what the client language is.*

```python
url = 'https://www.rottentomatoes.com/browse/movies_at_home/affiliates:netflix~sort:critic_highest'
page = requests.get(url, headers={'Accept-Language': "lang=en-US"})
```

```python
page.status_code
```

```python
soup = BeautifulSoup(page.content, "html.parser")
```

Using the Element Inspector tool, we can see that each movie title belongs to the `div` tag and is identified by the `class="js-tile-link"` parameter. Using the `.find_all()` method, we can find all those occurrences; we then save them to an object called `movies`.

```python
movies = soup.find_all('div', class_='flex-container')
```

```python
type(movies)
```

The `movies` object is a BeautifulSoup `ResultSet` object, which is an iterable and contains all the movie titles we're interested in. Let's explore the first element in the result set:

```python
movies[0]
```

We can see that each element of the `movies` object contains a series of `rt-text` tags, from which we can extract their text content. Let's use the `.find_all()` method to extract each one of these from the first element in the `movies` object:

*Note: in HTML, an `rt-text` is a custom HTML element, specific to the Rotten Tomatoes website. it is recognizable by the mandatory hyphen in its name, which is the W3C convention for custom elements.*

```python
movies[0].find('rt-text')
```

We're interested in the second `<rt-text>` tag, as that's the one that contains the movie title, so let's use the `[2]` index to isolate it; furthermore, as we now know, to extract its text content we need to use the `.text` attribute:

```python
movies[0].find_all('rt-text')
```

```python
movies[0].find_all('rt-text')[2]
```

```python
movies[0].find_all('rt-text')[2].text
```

Applying the `.strip()` method to this result, we finally get a cleaned up version of the movie title:

```python
movies[0].find_all('rt-text')[2].text.strip()
```

```python
movies[0].find('rt-text', {"context":"label", "slot":"audienceScore"}).text.strip()
# movies[0].find('rt-text', slot="audienceScore").text.strip()
```

What we have done so far is extracting the movie title from the first element in the `movies` object. In order to **extract all the movie titles**, let's cycle through the `movies` object, extract each movie title and save it to a new list that we'll name `movie_names`:

```python
movie_names = []
for m in movies:
    movie_names.append(m.find_all('rt-text')[2].text.strip())
```

```python
movie_names[0:5]
```

Now, using the same method, we proceed to extract the streaming start date (which is stored in the third `<rt-text>` element):

```python
movie_dates = []
for m in movies:
    movie_dates.append(m.find_all('rt-text')[3].text.strip())
```

```python
movie_dates[0:5]
```

At this point, we'd like to retrieve the audience and critic scores, which are stored in the `<score-pairs-deprecated>` tag; let's use the `.find()` method to isolate this specific tag:

```python
movies[0]
```

```python
movies[0].find('score-pairs-deprecated')
```

You may have noticed that the `score-pairs-deprecated` tag has an attribute called `audienceScore` and one called `criticsScore`; using these two attributes, we'll be able to to extract the corresponding scores that we're interested in.

To access this information, we'll use the following combination `.find('rt-text', {'slot': 'ATTR_NAME'}).get_text(strip=True)`, where:

- the `.find()` method is used to locate the `rt-text` tags with specific `slot` attributes (`audienceScore` and `criticsScore`) ,
- the `get_text(strip=True)` method is used to get the text content without any leading or trailing whitespace.

```python
movies[0].find('score-pairs-deprecated').find('rt-text', {'slot': 'audienceScore'}).get_text(strip=True)
```

```python
movies[0].find('score-pairs-deprecated').find('rt-text', {'slot': 'criticsScore'}).get_text(strip=True)
```

Let's loop through all elements in the `movies` object to extract the audience and critics scores:

```python
movie_scores_audience = []
movie_scores_critics = []
for m in movies:
    movie_scores_audience.append(m.find('score-pairs-deprecated').find('rt-text', {'slot': 'audienceScore'}).get_text(strip=True))
    movie_scores_critics.append(m.find('score-pairs-deprecated').find('rt-text', {'slot': 'criticsScore'}).get_text(strip=True))
```

```python
movie_scores_audience[0:5]
```

```python
movie_scores_critics[0:5]
```

Finally, let's combine together the four lists that we have created into a single DataFrame named `df_movies`:

```python
df_movies = pd.DataFrame(
    {'movie_name': movie_names,
     'streaming_date': movie_dates,
     'audience_score': movie_scores_audience,
     'critics_score': movie_scores_critics
    })
```

```python
df_movies
```

```python
df_movies.shape
```

One last thing that we can do is to clean the data frame by:

- formatting the `streaming_date` column so that it is interpreted as a date type;
- converting missing scores to a numpy NaN value (otherwise you won't be able to cast it as a float);
- remove the `%` symbol from the critics and audience scores;
- cast the two scores as integers;
- sort the data frame by the `critics_score` and then by the `audience_score`.

```python
# format as a date data type
df_movies['streaming_date'] = pd.to_datetime(df_movies['streaming_date'].str.replace('Streaming ', ''))
```

```python
# convert missing scores to a numpy NaN value
df_movies.loc[df_movies['audience_score']=='', 'audience_score'] = np.nan
df_movies.loc[df_movies['critics_score']=='', 'critics_score'] = np.nan
```

```python
df_movies['audience_score'] = df_movies['audience_score'].str.replace('%', '')
df_movies['critics_score'] = df_movies['critics_score'].str.replace('%', '')
```

```python
# convert the scores to floats
df_movies = df_movies.astype({'audience_score':'float', 'critics_score':'float'})
```

```python
# check data types
df_movies.info()
```

```python
# sorting data frame
df_movies.sort_values(['critics_score', 'audience_score'], ascending = False, inplace = True)
```

```python
df_movies
```
